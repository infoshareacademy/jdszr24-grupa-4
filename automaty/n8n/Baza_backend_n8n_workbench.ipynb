## 4. Dokumentacja workflowów n8n

Użyj tej sekcji, żeby dokumentować i projektować workflowy n8n równolegle z backendem.

Przykładowe bloki:
- **Workflow 1 – Chat na stronie**
  - Trigger: Webhook
  - Nody: Edit Fields → HTTP Request → HTML → Merge → Code → Edit Fields → AI Agent → Respond to Webhook

- **Workflow 2 – Logowanie Q&A do Postgresa**
  - Wejście: wynik Workflow 1
  - Nody: Edit Fields (rekord Q&A) → Postgres Insert


### 4.1. Workflow 1 – Chat na stronie (szkic)

- **Webhook**
  - wejście: `body.message`, `sessionId`, `pageUrl`, `source`.
- **Edit Fields**
  - `userMessage = body.message`.
- **HTTP Request**
  - pobiera HTML strony.
- **HTML node**
  - wyciąga treści (nagłówki, formularze, kontakt).
- **Merge**
  - łączy HTML + `userMessage`.
- **Code node**
  - czyści i strukturyzuje dane, ustala `Message input`.
- **Edit Fields1**
  - przekazuje `userMessage` dalej.
- **AI Agent**
  - generuje odpowiedź.
- **Respond to Webhook**
  - wysyła odpowiedź do frontu.


### 4.2. Workflow 2 – Logowanie Q&A do Postgresa (szkic)

- **Trigger**: Webhook / połączenie z Workflow 1.
- **Edit Fields (Build Q&A Record)**
  - `user_message = userMessage`
  - `assistant_answer = <pole z AI Agent>`
  - `session_id = sessionId`
  - `page_url = pageUrl`
  - `source = 'localhost'` lub inny kontekst
  - `intent = 'unknown'` (do późniejszej klasyfikacji)
- **Postgres → Insert**
  - zapis do tabeli `chat_qa`.
