# Nauka od zera: venv + VS Code + Jupyter Notebook

Cel: skonfigurować lokalne środowisko Python tylko dla jednego projektu i uruchamiać w nim pliki `.ipynb` w VS Code.

## KROK 1: Utwórz folder projektu i otwórz go w VS Code

1. Utwórz nowy folder projektu, np. `moj-projekt`.
2. Otwórz **dokładnie ten folder** w VS Code:
   - `File -> Open Folder`
   - wybierz folder projektu
3. Od tej chwili wszystkie kolejne komendy wykonuj w terminalu otwartym **w tym folderze**.

Ważne: środowisko `venv` ma być utworzone **wewnątrz repozytorium / folderu projektu**, żeby było przypisane tylko do niego.

## KROK 2: Utwórz środowisko wirtualne `venv` w folderze projektu

Otwórz terminal w VS Code:


- menu: `Terminal -> New Terminal`

Upewnij się, że jesteś w folderze projektu, a potem wykonaj jedną z poniższych komend.



### Windows PowerShell / CMD

```Terminal
python -m venv .venv
```


Jeśli `python` nie działa, spróbuj:

```Terminal
py -m venv .venv
```

### macOS / Linux

```Terminal
python3 -m venv .venv
```

Po wykonaniu komendy w folderze projektu pojawi się katalog `.venv`.

Przykład struktury:

```text
moj-projekt/
├─ .venv/
├─ notebook.ipynb
└─ ...
```


## KROK 3: Aktywuj środowisko w terminalu

Po aktywacji terminal zacznie używać Pythona i `pip` z tego projektu.

### Windows PowerShell

```Terminal
.\.venv\Scripts\Activate.ps1
```

### Windows CMD

```Terminal
.\.venv\Scripts\activate.bat
```

### macOS / Linux

```Terminal
source .venv/bin/activate
```

Po aktywacji zwykle zobaczysz w terminalu prefiks podobny do:

```text
(.venv)
```

To znak, że pracujesz już we właściwym środowisku.

### Instalacja w aktywnym środowisku

```Terminal
python -m pip install --upgrade pip
python -m pip install ipykernel pandas
```

## KROK 6: Test praktyczny

Uruchom poniższą komórkę Python w notebooku. Jeśli wszystko jest dobrze ustawione, import `pandas` zadziała, a ścieżka do interpretera będzie wskazywać na `.venv`.

````python
import sys
import pandas as pd

print("Interpreter:", sys.executable)
print("Wersja pandas:", pd.__version__)

df = pd.DataFrame({
    "produkt": ["krzesło", "stół", "lampa"],
    "cena": [120, 340, 89]
})

df["cena_z_vat_23"] = df["cena"] * 1.23
df
````

### Co sprawdzić po uruchomieniu

1. `Interpreter:` powinien wskazywać ścieżkę z folderem `.venv`
   - Windows: coś w stylu `.venv\Scripts\python.exe`
   - macOS/Linux: coś w stylu `.venv/bin/python`
2. `pandas` powinien importować się bez błędu.
3. Tabela `DataFrame` powinna się wyświetlić w notebooku.

To potwierdza, że notebook działa na interpreterze z Twojego lokalnego środowiska projektu, a nie na przypadkowym globalnym Pythonie.

````python
import sys
from pathlib import Path

python_path = Path(sys.executable)
print("Czy interpreter pochodzi z .venv?:", ".venv" in str(python_path))
````

## KROK 7: Dobra praktyka — `requirements.txt` i dezaktywacja środowiska

Na końcu zapisz listę bibliotek do pliku `requirements.txt`, żeby dało się łatwo odtworzyć środowisko na innym komputerze albo po sklonowaniu repozytorium.

### Zapis bibliotek

Uruchom w terminalu z aktywnym środowiskiem:

```Terminal
pip freeze > requirements.txt
```

Sprawdź zawartość:

```Terminal
cat requirements.txt
```

Na Windows CMD możesz użyć:

```Terminal
type requirements.txt
```

### Odtworzenie środowiska w przyszłości

Gdy projekt będzie już miał plik `requirements.txt`, odtworzysz środowisko tak:

```Terminal
python -m venv .venv
```

Aktywacja, a potem:

```Terminal
python -m pip install --upgrade pip
python -m pip install -r requirements.txt
```

### Dezaktywacja środowiska

Kiedy kończysz pracę w terminalu, wpisz:

```Terminal
deactivate
```
