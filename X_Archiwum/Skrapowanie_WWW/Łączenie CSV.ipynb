```python
import pandas as pd
from pathlib import Path
from urllib.parse import urlparse

# Ścieżki
base_path = Path(r"C:\1\T4\jdszr24-grupa-4\BAZA_CZAT\BAZA\DATA\faq_output.csv")
mebligo_path = Path(r"C:\1\T4\jdszr24-grupa-4\X_Archiwum\Skrapowanie_WWW\mebligo_faq.csv")
faq2_path = Path(r"C:\1\T4\jdszr24-grupa-4\X_Archiwum\Skrapowanie_WWW\faq_dataset.csv")

def root_url(url: str) -> str:
    """Wyciąga https://domena.tld z pełnego URL."""
    if not isinstance(url, str) or not url:
        return ""
    parsed = urlparse(url)
    scheme = parsed.scheme or "https"
    netloc = parsed.netloc
    return f"{scheme}://{netloc}" if netloc else url

# 1. Wczytanie bazy wzorcowej
base_df = pd.read_csv(base_path, sep=";")

# 2. Wczytanie Mebligo
meb = pd.read_csv(mebligo_path)

meb_norm = pd.DataFrame({
    "sklep_marka": meb["shop"],
    "domowy_url": meb["source_url"].apply(root_url),
    "kat_a": "",
    "kat_b": "",
    "pytanie": meb["question"],
    "odpowiedz": meb["answer"],
})

# 3. Wczytanie Selsey (faq_dataset)
faq2 = pd.read_csv(faq2_path)

selsey = faq2[faq2["shop"] == "Selsey"].copy()

sel_norm = pd.DataFrame({
    "sklep_marka": selsey["shop"],
    "domowy_url": selsey["source_url"].apply(root_url),
    "kat_a": "",
    "kat_b": "",
    "pytanie": selsey["question"],
    "odpowiedz": selsey["answer"],
})

# 4. Połączenie z bazą wzorcową
merged = pd.concat([base_df, meb_norm, sel_norm], ignore_index=True)

# 5. Zapis nadpisujący lub do nowego pliku
out_path = base_path  # lub np. base_path.with_name("faq_output_merged.csv")
merged.to_csv(out_path, sep=";", index=False, encoding="utf-8-sig")
print("Zapisano do:", out_path)
```
