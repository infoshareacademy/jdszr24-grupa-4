# Jak zamienić plik Markdown (.md) na skoroszyt Jupyter (.ipynb) w VS Code

Najszybszym i najwygodniejszym sposobem na zamianę pliku Markdown (`.md`) na skoroszyt Jupyter (`.ipynb`) bezpośrednio w Visual Studio Code jest użycie rozszerzenia Jupytext lub wbudowanego formatu "Percent format".

Oto dwa najprostsze sposoby, aby zrobić to w kilka sekund:

## Sposób 1: Przez format tekstowy "Percent" (Najszybszy, bez instalowania dodatków)

VS Code i rozszerzenie Jupyter domyślnie wspierają uruchamianie specjalnie sformatowanych plików tekstowych jako notebooków.

1. Otwórz swój plik `.md` w VS Code.
2. Na samej górze pliku dopisz:
   ```markdown
   #%%
   ```
3. Przed każdym nagłówkiem lub sekcją, którą chcesz zamienić w osobną komórkę, również dopisz `#%%`.
4. Jeśli chcesz, aby komórka zawierała kod, po prostu wpisz tam kod.
5. Jeśli komórka ma być Markdownem, dopisz `[markdown]`, np.:
   ```markdown
   #%% [markdown]
   # To jest nagłówek w komórce Markdown
   ```
6. Nad wpisanym `#%%` pojawi się mały napis "Run Cell" lub "Export".
7. Kliknij ikonę trzech kropek `...` w prawym górnym rogu edytora (lub kliknij prawym przyciskiem myszy w kodzie) i wybierz **Export -> Export as Jupyter Notebook**.

## Sposób 2: Użycie rozszerzenia Jupytext (Automatyczna konwersja)

Jeśli masz dużo plików lub nie chcesz ręcznie edytować tekstu, Jupytext zrobi to za Ciebie.
Upewnij się, że masz zainstalowany pakiet jupytext w swoim środowisku Python:

````python
!pip install jupytext
````

Otwórz terminal w VS Code (`Ctrl + \`` lub `Cmd + \`` na Macu). Uruchom jedną prostą komendę, która natychmiast stworzy plik `.ipynb` na bazie Twojego Markdowna:

````python
!jupytext --to notebook nazwa_pliku.md
````

W folderze pojawi się gotowy plik `nazwa_pliku.ipynb`, który możesz od razu otworzyć w VS Code.

## Sposób 3: Magiczna komórka wewnątrz VS Code

Jeśli masz już otwarty jakikolwiek pusty Jupyter Notebook w VS Code i chcesz po prostu "wciągnąć" do niego treść pliku Markdown:

1. Stwórz nową komórkę kodu w notebooku.
2. Wpisz i uruchom poniższe polecenie (zastępując ścieżkę do swojego pliku):

````python
!jupytext --to notebook moj_plik.md
````

3. Odśwież widok eksploratora plików – Twój nowy skoroszyt jest gotowy.
